```python
import requests
import pandas as pd
import sqlite3
```

```python
from google.colab import userdata
import requests

API_KEY = userdata.get('TMDB_API_KEY')

url = "https://api.themoviedb.org/3/discover/movie"

params = {
    "api_key": API_KEY
}

response = requests.get(url, params=params)
data = response.json()
```

```python
movies = data['results']

df = pd.DataFrame(movies)

# Select useful columns (optional but cleaner)
df = df[[
    'id', 'title', 'release_date', 'popularity',
    'vote_average', 'vote_count', 'genre_ids'
]]

df.head()
```

```python
df['genre_ids'] = df['genre_ids'].apply(lambda x: str(x))
```

```python
conn = sqlite3.connect("movies.db")

df.to_sql("movies", conn, if_exists="replace", index=False)

conn.close()
```

```python
conn = sqlite3.connect("movies.db")

movies_df = pd.read_sql("SELECT * FROM movies", conn)

conn.close()
```

```python
import ast
movies_df['genre_ids'] = movies_df['genre_ids'].apply(ast.literal_eval)
```

```python
genres_df = movies_df.explode('genre_ids')

genre_counts = genres_df['genre_ids'].value_counts()

genre_counts
```
